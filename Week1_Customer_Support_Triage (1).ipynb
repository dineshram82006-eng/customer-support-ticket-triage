```python
import pandas as pd

df_tickets = pd.read_csv('/content/customer_support_tickets.csv')

print(f"DataFrame has {df_tickets.shape[0]} rows and {df_tickets.shape[1]} columns.\n")

print("Missing values per column:")
print(df_tickets.isnull().sum())

print("\nFirst 5 rows of the DataFrame:")
print(df_tickets.head())
```

```python
df_tickets['Ticket Description'] = df_tickets['Ticket Description'].str.lower()
print("After converting to lowercase:")
print(df_tickets[['Ticket ID', 'Ticket Description']].head())

df_tickets['Ticket Description'] = df_tickets['Ticket Description'].str.replace('[^a-z0-9 ]', '', regex=True)
print("\nAfter removing special characters:")
print(df_tickets[['Ticket ID', 'Ticket Description']].head())

df_tickets['Ticket Description'] = df_tickets['Ticket Description'].str.strip().str.replace('\s+', ' ', regex=True)
print("\nAfter removing extra spaces:")
print(df_tickets[['Ticket ID', 'Ticket Description']].head())
```

```python
df_tickets['Ticket Description'] = df_tickets['Ticket Description'].str.lower()
print("After converting to lowercase:")
print(df_tickets[['Ticket ID', 'Ticket Description']].head())

df_tickets['Ticket Description'] = df_tickets['Ticket Description'].str.replace('[^a-z0-9 ]', '', regex=True)
print("\nAfter removing special characters:")
print(df_tickets[['Ticket ID', 'Ticket Description']].head())

df_tickets['Ticket Description'] = df_tickets['Ticket Description'].str.strip().str.replace(r'\s+', ' ', regex=True)
print("\nAfter removing extra spaces:")
print(df_tickets[['Ticket ID', 'Ticket Description']].head())
```

```python
keywords = {
    'PAYMENT': ['payment', 'bill', 'charge', 'invoice', 'credit card', 'transaction', 'billing', 'price', 'cost'],
    'LOGIN': ['login', 'account', 'password', 'access', 'credential', 'signin', 'username', 'locked', 'auth'],
    'DELIVERY': ['delivery', 'ship', 'tracking', 'order', 'package', 'shipping', 'arrive', 'received'],
    'REFUND': ['refund', 'return', 'money back', 'reimburse', 'cancel', 'exchange', 'warranty', 'guarantee'],
    'BUG': ['error', 'bug', 'crash', 'glitch', 'malfunction', 'broken', 'issue', 'not working', 'fix', 'problem'],
    'GENERAL': [] # Default category if no specific keywords are found
}

def classify_issue(description):
    description = str(description) # Ensure description is a string
    for issue_type, kw_list in keywords.items():
        if kw_list:
            for keyword in kw_list:
                if keyword in description:
                    return issue_type
    return 'GENERAL'

df_tickets['Issue Type'] = df_tickets['Ticket Description'].apply(classify_issue)

print("Count of tickets per Issue Type:")
print(df_tickets['Issue Type'].value_counts())
```

```python
priority_keywords = {
    'P0': ['urgent', 'critical', 'immediately', 'stop working', 'outage', 'severe', 'down', 'cannot access'],
    'P1': ['high', 'important', 'major', 'degraded', 'slow', 'security risk'],
    'P2': ['medium', 'moderate', 'normal', 'minor impact', 'bug', 'issue'],
    'P3': ['low', 'cosmetic', 'request', 'inquiry', 'question', 'suggestion']
}

def assign_priority(description):
    description = str(description)
    for keyword in priority_keywords['P0']:
        if keyword in description:
            return 'P0'
    for keyword in priority_keywords['P1']:
        if keyword in description:
            return 'P1'
    for keyword in priority_keywords['P2']:
        if keyword in description:
            return 'P2'
    for keyword in priority_keywords['P3']:
        if keyword in description:
            return 'P3'
    return 'P3'

df_tickets['Ticket Priority Level'] = df_tickets['Ticket Description'].apply(assign_priority)

print("Count of tickets per assigned Ticket Priority Level:")
print(df_tickets['Ticket Priority Level'].value_counts())
```

```python
df_tickets['Date of Purchase'] = pd.to_datetime(df_tickets['Date of Purchase'])

sla_mapping = {
    'P0': 4,
    'P1': 8,
    'P2': 24,
    'P3': 48
}

df_tickets['SLA Hours'] = df_tickets['Ticket Priority Level'].map(sla_mapping)

df_tickets['Due Time'] = df_tickets['Date of Purchase'] + pd.to_timedelta(df_tickets['SLA Hours'], unit='h')

print("First 5 rows with 'Ticket ID', 'Ticket Priority Level', 'Date of Purchase', 'SLA Hours', and 'Due Time':")
print(df_tickets[['Ticket ID', 'Ticket Priority Level', 'Date of Purchase', 'SLA Hours', 'Due Time']].head())
```

```python
print("Distribution of Issue Type:")
print(df_tickets['Issue Type'].value_counts())
```

```python
print("\nAverage SLA Hours per Ticket Priority Level:")
print(df_tickets.groupby('Ticket Priority Level')['SLA Hours'].mean())
```

```python
current_time = pd.Timestamp.now()

overdue_tickets = df_tickets[
    (df_tickets['Ticket Status'] != 'Closed') &
    (df_tickets['Due Time'] < current_time)
]

print(f"\nNumber of overdue tickets: {len(overdue_tickets)}")
```

```python
output_file_path = 'processed_customer_support_tickets.csv'
df_tickets.to_csv(output_file_path, index=False)

print(f"Processed data exported to '{output_file_path}' successfully.")
```
